# Linear A reader benchmark

Runs every reader in `benchmark/models.yaml` on the same held-out tablets and shows the results (issues #56–#59).

* Choose and tune models on **val**. Run **test** only for final candidates (`--split test --final`; each test run is logged).
* Needs the dataset in `fraction-solver/out/yolo_signs` (built with `python -m la.cli dataset --labels sign`) and the weights listed in `models.yaml`; models with missing weights are skipped.
* Everything runs on the CPU with 2 threads, like the web app.

```python
import os
import sys
from pathlib import Path

ROOT = Path.cwd() if (Path.cwd() / 'benchmark').is_dir() else Path.cwd().parent
os.chdir(ROOT); sys.path.insert(0, str(ROOT))
from benchmark import report, run
```

```python
# Add --only <names> to run a subset, or --limit 10 for a quick check.
out = run.main(['--split', 'val'])
```

```python
report.main([str(out)])
from IPython.display import Image, Markdown, display

display(Markdown((out / 'report.md').read_text(encoding='utf-8')))
for name in ('top1', 'coverage', 'frequency', 'detection'):
    if (out / f'{name}.png').exists():
        display(Image(filename=str(out / f'{name}.png')))
```
